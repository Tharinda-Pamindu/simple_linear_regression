```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics

sns.set_style('whitegrid')
```

```python
# Load data
df = pd.read_csv('advertising.csv')

print(df.head())
print(df.tail())

df.info()

print(f'Duplicate sample count: {df.duplicated().sum()}')

print(df.describe())
```

```python
print(df.ndim)
print(df.shape)
```

```python
print('\nPair plot for every data')
sns.pairplot(df)
plt.show()
```

```python
x = df[['TV']]
y = df['Sales']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
print(f'Training set shape : {x_train.shape}')
print(f'Test set shape : {x_test.shape}')
```

```python
simple_model = LinearRegression()
```

```python
simple_model.fit(x_train, y_train)

print('model completion successfully')
```

```python
print(f'Bias (Intercept) -> {simple_model.intercept_:.4f}')
print(f'Weight (coefficient) -> {simple_model.coef_[0]:.4f}')
```

```python
# RMSE for training and testing datasets

y_train_pred = simple_model.predict(x_train)
y_test_pred = simple_model.predict(x_test)

train_rmse = np.sqrt(metrics.mean_squared_error(y_train, y_train_pred))
test_rmse = np.sqrt(metrics.mean_squared_error(y_test, y_test_pred))

print(f"Train RMSE: {train_rmse:.4f}")
print(f"Test RMSE:  {test_rmse:.4f}")
```
